Architecture parameters

```python
eps=0.01
np.random.seed(0)
W1 = np.random.randn(3, 8) * eps    # row i = the 8 incoming weights of hidden unit i
b1 = np.random.randn(3) * eps
W2 = np.random.randn(8, 3) * eps    # row i = the 3 incoming weights of output unit i
b2 = np.random.randn(8) * eps
lr=1
n_iter=300000
```

Main Functions

```python
def forwardprop(x):
    a2=sigmoid(W1 @ x+b1)  #values of the 3 hidden nodes after activation, a2=matrix of them
    a3=sigmoid(W2 @ a2+b2) #values of the 8 output nodes after acivation, a3=matrix of them
    return a2,a3

def backprop(x,y):
    a2,a3=forwardprop(x)
    d3=-(y-a3) * (a3 * (1-a3)) #output layer error term, error * last term is the derivative of the sigmoid
    d2=(W2.T @ d3) * a2 * (1-a2) #hidden layer error term, same as above
    return np.outer(d2,x), d2, np.outer(d3, a2), d3

def gradient_descent(x,y, m, lr, n_iter, W1, b1, W2, b2):
    for it in range(n_iter):
        dW1=np.zeros_like(W1)
        db1=np.zeros_like(b1)
        dW2=np.zeros_like(W2)
        db2=np.zeros_like(b2)
        for i in range(m):
            gW1, gb1, gW2, gb2 = backprop(x[i],y[i])
            dW1 += gW1
            db1 += gb1
            dW2 += gW2
            db2 += gb2
        W1-=lr*dW1/m
        b1-=lr*db1/m
        W2-=lr*dW2/m
        b2-=lr*db2/m

gradient_descent(data,Y, m, lr, n_iter, W1, b1, W2, b2)

print("final cost:", cost())
for i in range(m):
    a2, a3 = forwardprop(data[i])
    print(data[i].astype(int), "hidden:", np.round(a2, 2), "output:", np.round(a3, 2))


```
